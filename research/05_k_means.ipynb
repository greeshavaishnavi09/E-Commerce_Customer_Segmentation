```python
# IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
```

```python
# LOAD TRANSFORMED DATA

rfm = pd.read_csv("../artifacts/data_transformation/customer_rfm.csv")

rfm.head()
```

```python
# VERIFY TRANSFORMED DATA

print("Shape:", rfm.shape)

print("\nColumns:")
print(rfm.columns.tolist())

print("\nData Types:")
print(rfm.dtypes)

print("\nMissing Values:")
print(rfm.isnull().sum())
```

```python
# VIEW TRANSFORMED RFM DATA

rfm.head(10)
```

```python
# SEPARATE CUSTOMER ID AND FEATURES

customer_ids = rfm["CustomerID"]

X = rfm[["Recency", "Frequency", "Monetary"]]

print("Customer IDs shape:", customer_ids.shape)
print("Feature shape:", X.shape)
```

```python
# Check skewness

X.skew()
```

```python
# Log transform RFM features

X_log = np.log1p(X)   # To reduce skewness and compress extreme values

X_log.head()
```

```python
# Check skewness after log transformation

X_log.skew()    # skewness to become less extreme
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# Scale transformed features

scaler = StandardScaler()  #To put the features on a comparable scale because K-Means is distance-based

X_scaled = scaler.fit_transform(X_log)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X_log.columns,
    index=X_log.index
)

X_scaled.head()
```

```python
# Check scaled data

print(X_scaled.mean())
print(X_scaled.std())
```

# Baseline Model

```python
from sklearn.cluster import KMeans
```

```python
inertia = []         #We'll store the inertia for every K

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    inertia.append(kmeans.inertia_)   #Inertia naturally decreases as K increases
```

```python
# Print inertia values

for k, value in zip(range(2, 11), inertia):
    print(f"K = {k}, Inertia = {value:.2f}")
```

# Elbow Method

```python
import matplotlib.pyplot as plt
```

```python
plt.plot(range(2, 11), inertia, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```

# Silhouette Score

```python
# Import silhouette score

from sklearn.metrics import silhouette_score
```

```python
# Calculate silhouette score for different K values

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_scaled)
    
    score = silhouette_score(X_scaled, labels)
    
    silhouette_scores.append(score)
```

```python
# Print silhouette scores

for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
# Silhouette Method

plt.plot(range(2, 11), silhouette_scores, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score")

plt.show()
```

# Compare K Values

```python
# Find the K with the highest silhouette score

best_k = range(2, 11)[np.argmax(silhouette_scores)]

best_score = max(silhouette_scores)

print("Best K:", best_k)
print("Best Silhouette Score:", round(best_score, 4))
```

# Final K-Means Model

```python
# Train final K-Means model

final_kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

cluster_labels = final_kmeans.fit_predict(X_scaled)
```

```python
# Add cluster labels to RFM data

rfm["Cluster"] = cluster_labels

rfm.head()
```

```python
# Check cluster distribution

print(rfm["Cluster"].value_counts().sort_index())
```

```python
# Calculate cluster percentage

cluster_percentage = (
    rfm["Cluster"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
)

cluster_percentage
```
