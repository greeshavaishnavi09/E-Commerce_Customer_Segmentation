```python
# IMPORT REQUIRED LIBRARIES

import pandas as pd
import numpy as np
```

```python
# LOAD TRANSFORMED DATA

rfm = pd.read_csv("../artifacts/data_transformation/customer_rfm.csv")

rfm.head()
```

```python
# VERIFY TRANSFORMED DATA

print("Shape:", rfm.shape)

print("\nColumns:")
print(rfm.columns.tolist())

print("\nData Types:")
print(rfm.dtypes)

print("\nMissing Values:")
print(rfm.isnull().sum())
```

```python
# VIEW TRANSFORMED RFM DATA

rfm.head(10)
```

```python
# SEPARATE CUSTOMER ID AND FEATURES

customer_ids = rfm["CustomerID"]

X = rfm[["Recency", "Frequency", "Monetary"]]

print("Customer IDs shape:", customer_ids.shape)
print("Feature shape:", X.shape)
```

```python
# Check skewness

X.skew()
```

```python
# Log transform RFM features

X_log = np.log1p(X)   # To reduce skewness and compress extreme values

X_log.head()
```

```python
# Check skewness after log transformation

X_log.skew()    # skewness to become less extreme
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
# Scale transformed features

scaler = StandardScaler()  #To put the features on a comparable scale because K-Means is distance-based

X_scaled = scaler.fit_transform(X_log)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X_log.columns,
    index=X_log.index
)

X_scaled.head()
```

```python
# Check scaled data

print(X_scaled.mean())
print(X_scaled.std())
```

# Baseline Model

```python
from sklearn.cluster import KMeans
```

```python
inertia = []         #We'll store the inertia for every K

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    kmeans.fit(X_scaled)
    
    inertia.append(kmeans.inertia_)   #Inertia naturally decreases as K increases
```
