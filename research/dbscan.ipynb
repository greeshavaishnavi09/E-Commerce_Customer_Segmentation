```python
# Import libraries

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Check current working directory

print(os.getcwd())
```

```python
# Load transformed data

rfm = pd.read_csv("../artifacts/data_transformation/customer_rfm.csv")

rfm.head()
```

```python
# Check data shape

print("Shape:", rfm.shape)
```

```python
# Check columns

print("Columns:", rfm.columns.tolist())
```

```python
# Check missing values

print(rfm.isnull().sum())
```

```python
# Separate customer ID and features

customer_ids = rfm["CustomerID"]

X = rfm[["Recency", "Frequency", "Monetary"]]

print("Customer IDs shape:", customer_ids.shape)
print("Features shape:", X.shape)
```

# Data Preprocessing

```python
# Check skewness

print(X.skew())
```

```python
# Apply log transformation

X_log = np.log1p(X)

X_log.head()
```

```python
# Check skewness after log transformation

print(X_log.skew())
```

```python
# Import StandardScaler

from sklearn.preprocessing import StandardScaler
```

```python
# Scale features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X_log)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X_log.columns,
    index=X_log.index
)

X_scaled.head()
```

```python
# Check scaled features

print("Mean:")
print(X_scaled.mean())

print("\nStandard Deviation:")
print(X_scaled.std())
```
